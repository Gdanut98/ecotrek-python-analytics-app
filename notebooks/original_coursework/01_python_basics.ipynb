```python
# Create Variables using Constants Provided
Unit_Cost = 35
August_Sales = 3000 # Projected number of units sold for August
```

```python
# Calculate Monthly Projected Revenue
August_Revenue = Unit_Cost * August_Sales # Calculate August Revenue with the given above information
September_Revenue = (August_Revenue * .15) + August_Revenue # Using August Revenue, calculated a projected 15% increase for Septemeber Revenue
October_Revenue = (September_Revenue * .20) + September_Revenue # Using September Revenue, calculated a projected 20% for October Revenue
```

```python
# Calculate Total 3rd Quarter Revenue
Q3_Projected_Revenue = August_Revenue + September_Revenue + October_Revenue
```

```python
# Print Monthly Revenue Projections and Total Q3 Projections using F strings
print(" Quarterly Sales Revenue Projection ")
print(f"August Projected Revenue: ${August_Revenue:,.2f}")
print(f"September Projected Revenue: ${September_Revenue:,.2f}")
print(f"October Projected Revenue: ${October_Revenue:,.2f}")
print(f"Total Q3 Projected Revenue is: ${Q3_Projected_Revenue:,.2f}")
```
