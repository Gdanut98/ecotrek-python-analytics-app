```python
# Import neccessary packages

import numpy as np
import pandas as pd
import scipy.stats
from scipy.stats import pearsonr

pd.set_option('display.max_columns', None)
```

```python
# Create a DataFrame to read in data from provided file

df = pd.read_csv('Week_3_Temperature_DailySale.csv')
```

```python
# Rename columns to more user friendly names

df.rename(columns = {'Daily Units Sold': 'daily_units_sold',
                     'Daily Temperature (C)' : 'daily_temp'
                    },
          inplace = True
         )
```

```python
# Conduct Pearson Correlation to determine relationship between daily sold units and daily temperature

r, p_value = pearsonr(df['daily_units_sold'], df['daily_temp'])
print(f"Pearson r: {r}, p-value: {p_value}") 
```

## The results indicate a strong positive relationship between daily sold units and daily temperature

```python
# Sort the dataframe by daily units sold assuming descending

df.sort_values(by='daily_units_sold',ascending=False, inplace = True)
```

```python
# Write the sorted dataframe in a new csv file

df.to_csv("GreenTote_analysis_Danut_George.csv", index=False)
```

```python
# Create a DataFrame to read in proced review file

reviews = pd.read_csv('Backpack_Reviews_with_Ratings.csv')
reviews.columns = reviews.columns.str.strip() # Clean the columns just in case
reviews.head() # review first few rows
```

```python
# Import neccessary packages for SentimentIntensityAnalyzer

!pip install vaderSentiment
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

```python
# Use SentimentIntensityAnalyze and for loops to loop through each review and give it a score and sentiment

# Initialize the Sentiment Intensity Analyzer
sia = SentimentIntensityAnalyzer()

# Empty list to hold sentiment labels
sentiments = []  

# Loop through each review
for review in reviews['Review']:
    score = sia.polarity_scores(str(review))['compound']
    if score >= 0.05:
        sentiments.append("positive")
    elif score <= -0.05:
        sentiments.append("negative")
    else:
        sentiments.append("neutral")

# Add new column to the same DataFrame
reviews['sentiment'] = sentiments

print(reviews)
```

```python
# Calculate the average rating og the reviews for each rating

avg_rating = reviews.groupby('sentiment')['Rating'].mean()
```

```python
# Calculate the total number of reviews by sentiment
total_rating = reviews['sentiment'].value_counts()
```

```python
# Display Results

print("Average Ratings by Sentiment:")
for sentiment, rating in avg_rating.items():
    print(f"{sentiment.capitalize():<8}: {rating:.2f}")

print("Total Amount of Ratngs by Sentiment:")
for sentiment, rating in total_rating.items():
    print(f"{sentiment.capitalize():<8}: {rating:.2f}")
```

```python
# Write the reviews dataframe in a new csv file

reviews.to_csv("Danut_George_sentiment.csv", index=False)
```
